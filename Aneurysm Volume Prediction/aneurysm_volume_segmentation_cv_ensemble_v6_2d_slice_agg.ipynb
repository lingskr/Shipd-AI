# Aneurysm Volume Prediction v6 (2D Slice Segmentation + Depth Aggregation)

本版本严格按你的5条标准设计：
1. 分割建模而非回归
2. 数据增强（翻转/旋转/强度扰动）
3. 分割损失（BCE + Dice）
4. 基于NIfTI spacing计算体积
5. 2D切片推理并沿深度维度做原则性聚合

说明：这是基于你高分路线（v2_attention_plus=0.7976）的“稳健2D版”。

```python
def load_nii(path):
    nii = nib.load(str(path))
    arr = nii.get_fdata(dtype=np.float32)
    spacing = nii.header.get_zooms()[:3]
    return arr, spacing

def robust_zscore(x, eps=1e-6):
    lo, hi = np.percentile(x, [0.5, 99.5])
    x = np.clip(x, lo, hi)
    m, s = x.mean(), x.std()
    return (x - m) / (s + eps)

def parse_pid(path):
    return int(Path(path).name.split(".")[0])

def volume_from_binary(mask_dhw, spacing):
    voxel = float(spacing[0] * spacing[1] * spacing[2])
    return float(max(mask_dhw.sum() * voxel, 0.0))

def volume_from_prob(prob_dhw, spacing):
    voxel = float(spacing[0] * spacing[1] * spacing[2])
    return float(max(prob_dhw.sum() * voxel, 0.0))

def volumetric_similarity(y_true, y_pred, eps=1e-4):
    y_true = np.asarray(y_true, dtype=np.float64)
    y_pred = np.asarray(y_pred, dtype=np.float64)
    return float(np.mean(1.0 - np.abs(y_true - y_pred) / (y_true + y_pred + eps)))

def depth_aggregate(prob_dhw):
    # 原则性深度聚合：沿深度做固定核平滑，抑制孤立切片噪声
    # prob_dhw: (D,H,W)
    D = prob_dhw.shape[0]
    out = np.zeros_like(prob_dhw, dtype=np.float32)
    for z in range(D):
        z0 = max(0, z-1)
        z1 = z
        z2 = min(D-1, z+1)
        out[z] = 0.25 * prob_dhw[z0] + 0.5 * prob_dhw[z1] + 0.25 * prob_dhw[z2]
    return out
```

```python
class CaseCache:
    def __init__(self):
        self.img = {}
        self.msk = {}
        self.spacing = {}

    def load_case(self, img_path, mask_path=None):
        pid = parse_pid(img_path)
        if pid not in self.img:
            arr, sp = load_nii(img_path)
            arr = robust_zscore(arr)
            # (H,W,D) -> (D,H,W)
            arr = np.transpose(arr, (2,0,1)).astype(np.float32)
            self.img[pid] = arr
            self.spacing[pid] = np.asarray(sp, dtype=np.float32)
        if mask_path is not None and pid not in self.msk:
            m, _ = load_nii(mask_path)
            m = (m > 0.5).astype(np.float32)
            m = np.transpose(m, (2,0,1)).astype(np.float32)
            self.msk[pid] = m
        return pid

cache = CaseCache()

class SliceDataset2p5D(Dataset):
    def __init__(self, img_paths, mask_paths=None, augment=False):
        self.augment = augment
        self.samples = []  # (pid, z)
        self.has_mask = mask_paths is not None
        if self.has_mask:
            for ip, mp in zip(img_paths, mask_paths):
                pid = cache.load_case(ip, mp)
                D = cache.img[pid].shape[0]
                for z in range(D):
                    self.samples.append((pid, z))
        else:
            for ip in img_paths:
                pid = cache.load_case(ip, None)
                D = cache.img[pid].shape[0]
                for z in range(D):
                    self.samples.append((pid, z))

    def __len__(self):
        return len(self.samples)

    def _get_2p5d(self, vol, z):
        D = vol.shape[0]
        z0 = max(0, z-1)
        z1 = z
        z2 = min(D-1, z+1)
        x = np.stack([vol[z0], vol[z1], vol[z2]], axis=0).astype(np.float32)
        return x

    def _aug(self, x, y):
        # x: (3,H,W), y:(1,H,W)
        if random.random() < 0.5:
            x = np.flip(x, axis=1).copy()
            if y is not None: y = np.flip(y, axis=1).copy()
        if random.random() < 0.5:
            x = np.flip(x, axis=2).copy()
            if y is not None: y = np.flip(y, axis=2).copy()
        if random.random() < 0.5:
            k = random.randint(0, 3)
            x = np.rot90(x, k=k, axes=(1,2)).copy()
            if y is not None: y = np.rot90(y, k=k, axes=(1,2)).copy()
        if random.random() < 0.7:
            x = x * (1.0 + random.uniform(-0.12, 0.12)) + random.uniform(-0.10, 0.10)
        return x, y

    def __getitem__(self, idx):
        pid, z = self.samples[idx]
        vol = cache.img[pid]  # (D,H,W)
        x = self._get_2p5d(vol, z)
        y = None
        if self.has_mask:
            y = cache.msk[pid][z][None, ...].astype(np.float32)
        if self.augment:
            x, y = self._aug(x, y)
        out = {"image": torch.from_numpy(x), "pid": pid, "z": z}
        if y is not None:
            out["mask"] = torch.from_numpy(y)
        return out
```

```python
def gn(ch):
    g = 8
    while ch % g != 0 and g > 1:
        g -= 1
    return nn.GroupNorm(g, ch)

class SE2D(nn.Module):
    def __init__(self, ch, r=8):
        super().__init__()
        hid = max(ch // r, 4)
        self.net = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Conv2d(ch, hid, 1), nn.ReLU(inplace=True), nn.Conv2d(hid, ch, 1), nn.Sigmoid())
    def forward(self, x):
        return x * self.net(x)

class ResBlock2D(nn.Module):
    def __init__(self, in_ch, out_ch, drop=0.05):
        super().__init__()
        self.c1 = nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False)
        self.n1 = gn(out_ch)
        self.c2 = nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False)
        self.n2 = gn(out_ch)
        self.se = SE2D(out_ch)
        self.drop = nn.Dropout2d(drop)
        self.act = nn.ReLU(inplace=True)
        self.proj = nn.Conv2d(in_ch, out_ch, 1, bias=False) if in_ch != out_ch else nn.Identity()
    def forward(self, x):
        i = self.proj(x)
        x = self.act(self.n1(self.c1(x)))
        x = self.n2(self.c2(x))
        x = self.se(x)
        x = self.drop(x)
        return self.act(x + i)

class UNet2D(nn.Module):
    def __init__(self, in_ch=3, base=32):
        super().__init__()
        self.e1 = ResBlock2D(in_ch, base)
        self.p1 = nn.MaxPool2d(2)
        self.e2 = ResBlock2D(base, base*2)
        self.p2 = nn.MaxPool2d(2)
        self.e3 = ResBlock2D(base*2, base*4)
        self.p3 = nn.MaxPool2d(2)
        self.b = ResBlock2D(base*4, base*8, drop=0.1)
        self.u3 = nn.ConvTranspose2d(base*8, base*4, 2, 2)
        self.d3 = ResBlock2D(base*8, base*4)
        self.u2 = nn.ConvTranspose2d(base*4, base*2, 2, 2)
        self.d2 = ResBlock2D(base*4, base*2)
        self.u1 = nn.ConvTranspose2d(base*2, base, 2, 2)
        self.d1 = ResBlock2D(base*2, base)
        self.head = nn.Conv2d(base, 1, 1)
    def forward(self, x):
        e1 = self.e1(x); e2 = self.e2(self.p1(e1)); e3 = self.e3(self.p2(e2)); b = self.b(self.p3(e3))
        d3 = self.u3(b); d3 = self.d3(torch.cat([d3, e3], dim=1))
        d2 = self.u2(d3); d2 = self.d2(torch.cat([d2, e2], dim=1))
        d1 = self.u1(d2); d1 = self.d1(torch.cat([d1, e1], dim=1))
        return self.head(d1)

class DiceBCELoss(nn.Module):
    def __init__(self, pos_weight=4.0, bce_w=0.45, smooth=1e-5):
        super().__init__()
        self.pos_weight = pos_weight
        self.bce_w = bce_w
        self.smooth = smooth
    def forward(self, logits, target):
        pw = torch.tensor([self.pos_weight], device=logits.device, dtype=logits.dtype)
        bce = nn.functional.binary_cross_entropy_with_logits(logits, target, pos_weight=pw)
        p = torch.sigmoid(logits).reshape(logits.size(0), -1)
        t = target.reshape(target.size(0), -1)
        inter = (p*t).sum(dim=1)
        den = p.sum(dim=1) + t.sum(dim=1)
        dice = 1.0 - (2.0*inter + self.smooth) / (den + self.smooth)
        return self.bce_w * bce + (1.0 - self.bce_w) * dice.mean()
```

```python
train_df = pd.read_csv(TRAIN_CSV)
train_df["patient_id"] = train_df["patient_id"].astype(int)
all_train_imgs = sorted(TRAIN_IMG_DIR.glob("*.nii.gz"))
all_train_msks = [TRAIN_MASK_DIR / p.name for p in all_train_imgs]
all_test_imgs = sorted(TEST_IMG_DIR.glob("*.nii.gz"))
print("train:", len(all_train_imgs), "test:", len(all_test_imgs))
```

```python
CFG = {
    "n_splits": 5,
    "epochs": 50,
    "batch_size": 32,
    "num_workers": 0,
    "lr": 1e-3,
    "weight_decay": 1e-4,
    "base_ch": 32,
    "grad_clip": 1.0,
    "tta": True,
    "early_stop_patience": 8,
    "seeds": [42, 2025],
}
print(CFG)
```

```python
def make_splits(train_df, n_splits=5, seed=42):
    vols = train_df.sort_values("patient_id")["volume"].values
    try:
        bins = pd.qcut(vols, q=5, labels=False, duplicates="drop")
    except Exception:
        bins = pd.cut(vols, bins=5, labels=False)
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    return list(skf.split(np.arange(len(all_train_imgs)), bins))

def predict_case_prob_2d(model, vol_dhw, batch_size=64, tta=True):
    # vol_dhw: (D,H,W), return prob_dhw
    D,H,W = vol_dhw.shape
    xs=[]
    for z in range(D):
        z0=max(0,z-1); z1=z; z2=min(D-1,z+1)
        x=np.stack([vol_dhw[z0], vol_dhw[z1], vol_dhw[z2]], axis=0).astype(np.float32)
        xs.append(x)
    xs=np.stack(xs, axis=0)  # (D,3,H,W)
    probs=[]
    model.eval()
    with torch.no_grad():
        for i in range(0, D, batch_size):
            xb = torch.from_numpy(xs[i:i+batch_size]).to(DEVICE)
            if tta:
                p0 = torch.sigmoid(model(xb))
                p1 = torch.sigmoid(model(torch.flip(xb, dims=[2]))); p1 = torch.flip(p1, dims=[2])
                p2 = torch.sigmoid(model(torch.flip(xb, dims=[3]))); p2 = torch.flip(p2, dims=[3])
                pb = (p0 + p1 + p2) / 3.0
            else:
                pb = torch.sigmoid(model(xb))
            probs.append(pb.cpu().numpy())
    probs=np.concatenate(probs, axis=0)[:,0]  # (D,H,W)
    return probs.astype(np.float32)

def train_one_fold(seed, fold, tr_idx, va_idx):
    random.seed(seed + fold)
    np.random.seed(seed + fold)
    torch.manual_seed(seed + fold)
    tr_imgs = [all_train_imgs[i] for i in tr_idx]
    tr_msks = [all_train_msks[i] for i in tr_idx]
    va_imgs = [all_train_imgs[i] for i in va_idx]
    va_msks = [all_train_msks[i] for i in va_idx]

    tr_ds = SliceDataset2p5D(tr_imgs, tr_msks, augment=True)
    tr_ld = DataLoader(tr_ds, batch_size=CFG["batch_size"], shuffle=True, num_workers=CFG["num_workers"])

    model = UNet2D(in_ch=3, base=CFG["base_ch"]).to(DEVICE)
    crit = DiceBCELoss(pos_weight=4.0, bce_w=0.45)
    opt = torch.optim.AdamW(model.parameters(), lr=CFG["lr"], weight_decay=CFG["weight_decay"])
    sch = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=CFG["epochs"], eta_min=1e-5)
    scaler = torch.cuda.amp.GradScaler(enabled=USE_AMP)

    sd = CKPT_DIR / f"seed_{seed}"
    sd.mkdir(parents=True, exist_ok=True)
    ck = sd / f"fold_{fold}.pt"
    best_vs, no_imp = -1.0, 0

    for ep in range(CFG["epochs"]):
        model.train(); losses=[]
        for b in tr_ld:
            x = b["image"].to(DEVICE)
            y = b["mask"].to(DEVICE)
            opt.zero_grad(set_to_none=True)
            with torch.cuda.amp.autocast(enabled=USE_AMP):
                logit = model(x)
                loss = crit(logit, y)
            scaler.scale(loss).backward()
            scaler.unscale_(opt)
            torch.nn.utils.clip_grad_norm_(model.parameters(), CFG["grad_clip"])
            scaler.step(opt); scaler.update()
            losses.append(loss.item())
        sch.step()

        # 验证：case级体积VS（2D切片->深度聚合）
        model.eval(); gts=[]; prs=[]
        with torch.no_grad():
            for ip, mp in zip(va_imgs, va_msks):
                pid = parse_pid(ip)
                vol = cache.img[pid]
                gt = cache.msk[pid]
                sp = cache.spacing[pid]
                prob = predict_case_prob_2d(model, vol, batch_size=64, tta=False)
                prob = depth_aggregate(prob)
                pm = (prob > 0.5).astype(np.uint8)
                gts.append(volume_from_binary(gt, sp))
                prs.append(volume_from_binary(pm, sp))
        vs = volumetric_similarity(gts, prs)
        print(f"seed {seed} fold {fold} ep {ep+1:02d}/{CFG["epochs"]} loss={np.mean(losses):.4f} valVS={vs:.4f}")
        if vs > best_vs:
            best_vs = vs; no_imp = 0; torch.save(model.state_dict(), ck)
        else:
            no_imp += 1
        if no_imp >= CFG["early_stop_patience"]:
            print(f"early stop seed {seed} fold {fold}, best={best_vs:.4f}")
            break
    return ck, best_vs
```

```python
start=time.time()
splits = make_splits(train_df, n_splits=CFG["n_splits"], seed=42)
model_paths_by_seed = {}
model_weights = {}
rows=[]

for seed in CFG["seeds"]:
    paths=[]
    for fold, (tr_idx, va_idx) in enumerate(splits):
        p, s = train_one_fold(seed, fold, tr_idx, va_idx)
        paths.append(p)
        model_weights[(seed, fold)] = max(float(s), 1e-6)
        rows.append({"seed":seed, "fold":fold, "best_vs":float(s), "ckpt":str(p)})
    model_paths_by_seed[seed] = paths

score_df = pd.DataFrame(rows)
score_df.to_csv(LOG_DIR / "fold_scores.csv", index=False)
print(score_df)
print(f"training done in {(time.time()-start)/60:.1f} min")
```

```python
def get_fold_ensemble_prob_for_case(pid, fold):
    # 同一fold下，跨seed加权平均
    vol = cache.img[pid]
    probs=[]; ws=[]
    for seed, plist in model_paths_by_seed.items():
        m = UNet2D(in_ch=3, base=CFG["base_ch"]).to(DEVICE)
        m.load_state_dict(torch.load(plist[fold], map_location=DEVICE))
        m.eval()
        pr = predict_case_prob_2d(m, vol, batch_size=64, tta=CFG["tta"])
        probs.append(pr)
        ws.append(model_weights[(seed, fold)])
    probs = np.stack(probs, axis=0)
    ws = np.asarray(ws, dtype=np.float64)
    ws = ws / (ws.sum() + 1e-12)
    out = np.tensordot(ws, probs, axes=(0,0)).astype(np.float32)
    return out

# ===== OOF调参：阈值 + 软硬融合（严格基于2D深度聚合） =====
gt_map = {int(r.patient_id): float(r.volume) for _, r in train_df.iterrows()}
oof_items=[]
for fold, (_, va_idx) in enumerate(splits):
    for i in va_idx:
        pid = parse_pid(all_train_imgs[i])
        sp = cache.spacing[pid]
        prob = get_fold_ensemble_prob_for_case(pid, fold)
        prob = depth_aggregate(prob)
        oof_items.append({"pid": pid, "spacing": sp, "prob": prob})

thr_grid = np.arange(0.30, 0.71, 0.01)
alpha_grid = np.arange(0.0, 1.01, 0.1)  # pred = alpha*hard + (1-alpha)*soft
best = {"vs": -1.0, "thr": 0.5, "alpha": 1.0}

for thr in thr_grid:
    hard=[]; soft=[]; y=[]
    for it in oof_items:
        pid = int(it["pid"])
        sp = it["spacing"]
        prob = it["prob"]
        hard.append(volume_from_binary((prob > thr).astype(np.uint8), sp))
        soft.append(volume_from_prob(prob, sp))
        y.append(gt_map[pid])
    hard=np.asarray(hard, dtype=np.float64)
    soft=np.asarray(soft, dtype=np.float64)
    y=np.asarray(y, dtype=np.float64)
    for a in alpha_grid:
        pred = a*hard + (1-a)*soft
        vs = volumetric_similarity(y, pred)
        if vs > best["vs"]:
            best = {"vs": float(vs), "thr": float(thr), "alpha": float(a)}

print("Best OOF before calibration:", best)

# 线性校准
rows=[]
for it in oof_items:
    pid = int(it["pid"])
    sp = it["spacing"]
    prob = it["prob"]
    vh = volume_from_binary((prob > best["thr"]).astype(np.uint8), sp)
    vs = volume_from_prob(prob, sp)
    blend = best["alpha"] * vh + (1-best["alpha"]) * vs
    rows.append({"patient_id": pid, "gt": gt_map[pid], "blend": blend})
oof_df = pd.DataFrame(rows).sort_values("patient_id").reset_index(drop=True)
k, b = np.polyfit(oof_df["blend"].values, oof_df["gt"].values, deg=1)
oof_df["calibrated"] = np.clip(k * oof_df["blend"].values + b, 0, None)
vs_uncal = volumetric_similarity(oof_df["gt"].values, oof_df["blend"].values)
vs_cal = volumetric_similarity(oof_df["gt"].values, oof_df["calibrated"].values)
use_cal = bool(vs_cal >= vs_uncal)
print(f"OOF VS uncal={vs_uncal:.6f} | cal={vs_cal:.6f} | use_cal={use_cal}")
oof_df.to_csv(PRED_DIR / "oof_predictions.csv", index=False)
```

```python
# ===== 测试集推理：跨seed+fold全模型融合（2D切片+深度聚合） =====
models=[]
weights=[]
for seed, plist in model_paths_by_seed.items():
    for fold, p in enumerate(plist):
        m = UNet2D(in_ch=3, base=CFG["base_ch"]).to(DEVICE)
        m.load_state_dict(torch.load(p, map_location=DEVICE))
        m.eval()
        models.append(m)
        weights.append(model_weights[(seed, fold)])
weights = np.asarray(weights, dtype=np.float64)
weights = weights / (weights.sum() + 1e-12)

rows=[]
for ip in tqdm(all_test_imgs, desc="Test inference"):
    pid = parse_pid(ip)
    cache.load_case(ip, None)
    vol = cache.img[pid]
    sp = cache.spacing[pid]
    probs=[]
    with torch.no_grad():
        for m in models:
            pr = predict_case_prob_2d(m, vol, batch_size=64, tta=CFG["tta"])
            probs.append(pr)
    probs = np.stack(probs, axis=0)
    prob = np.tensordot(weights, probs, axes=(0,0)).astype(np.float32)
    prob = depth_aggregate(prob)

    hard_vol = volume_from_binary((prob > best["thr"]).astype(np.uint8), sp)
    soft_vol = volume_from_prob(prob, sp)
    blend = best["alpha"] * hard_vol + (1-best["alpha"]) * soft_vol
    pred = np.clip(k * blend + b, 0, None) if use_cal else max(blend, 0.0)
    rows.append({"patient_id": pid, "volume": float(pred)})

sub = pd.DataFrame(rows).sort_values("patient_id").reset_index(drop=True)
sub.to_csv(PRED_DIR / "submission.csv", index=False)

targets = [
    Path("/root/setup/solution/working/submission.csv"),
    Path("/working/submission.csv"),
    OUTPUT_ROOT / "submission.csv",
]
saved=[]
for t in targets:
    try:
        t.parent.mkdir(parents=True, exist_ok=True)
        sub.to_csv(t, index=False)
        saved.append(str(t))
    except Exception as e:
        print("skip", t, e)

meta = {
    "run_name": RUN_NAME,
    "cfg": CFG,
    "best_oof": best,
    "calibration": {"use": use_cal, "k": float(k), "b": float(b), "vs_uncal": float(vs_uncal), "vs_cal": float(vs_cal)},
    "saved_submission_paths": saved,
}
with open(LOG_DIR / "run_meta.json", "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

print("Saved submission paths:")
for p in saved:
    print(" -", p)
print(sub.head())
```
